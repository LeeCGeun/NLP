# Construct Vocab(word level) with Text Preprocessing

```python
from tqdm import tqdm
from nltk.tokenize import word_tokenize
```

```python
import nltk

nltk.download("punkt")
```

```python
def construct_vocab(dataset):
    ## start ##
    vocab = {}
    for sentence in dataset:
        words = word_tokenize(sentence)
        for word in words:
            vocab[word] = vocab.get(word, 0) + 1
    return vocab

def vocab_analysis(vocab):
    print("Vocab size: ", len(vocab))
    print("Vocab sample: ", list(vocab.items())[:10])
```

### 0. None

```python
vocab_none = construct_vocab(dataset)

vocab_analysis(vocab_none)
```

### 1. Lower case

```python
def lowercase(dataset):
    ## start ##
    ret = [sentence.lower() for sentence in dataset]

    ### end ###

    return ret
```

```python
dataset_lowercase = lowercase(dataset)
vocab_lowercase = construct_vocab(dataset_lowercase)

vocab_analysis(vocab_lowercase)
```

### 2. Stopword Removal

```python
import nltk
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from nltk.corpus import stopwords

nltk.download("punkt")
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('stopwords')

example_words = ["plays","played","play","playing"]
test_sentence = [
    "I played and started playing with players in the playgrounds",
    "I am playing with my playmates"
    ]
```

```python
stop_words = set(stopwords.words('english'))  # 영어 불용어 리스트 가져오기

def remove_stopwords(dataset):
    global stop_words

    ## start ##
    new_dataset = [' '.join([word for word in word_tokenize(sentence) if word.lower() not in stop_words]) for sentence in dataset]

    ### end ###

    return new_dataset

print("Before stopword removal: ", test_sentence)
print("After stopword removal: ", remove_stopwords(test_sentence))
```

```python
dataset_remove_stopwords = remove_stopwords(dataset)
vocab_remove_stopwords = construct_vocab(dataset_remove_stopwords)

vocab_analysis(vocab_remove_stopwords)
```

### 3. Stemming

```python
ps = PorterStemmer()

print("{0:20}{1:20}".format("--Word--","--Stem--"))
for word in example_words:
   print ("{0:20}{1:20}".format(word, ps.stem(word)))
```

```python
def stemming(dataset):
    global ps

    ## start ##
    new_dataset = [' '.join([ps.stem(word) for word in word_tokenize(sentence)]) for sentence in dataset]
    ### end ###

    return new_dataset

print("Before stemming: ", test_sentence)
print("After stemming: ", stemming(test_sentence))
```

```python
dataset_stemming = stemming(dataset)
vocab_stemming = construct_vocab(dataset_stemming)

vocab_analysis(vocab_stemming)
```

### 4. Lemmatization

```python
def get_wordnet_pos(word):
    tag = nltk.pos_tag([word])[0][1][0].upper()
    tag_dict = {"J": wordnet.ADJ, "N": wordnet.NOUN, "V": wordnet.VERB, "R": wordnet.ADV}
    return tag_dict.get(tag, wordnet.NOUN)
```

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
le = WordNetLemmatizer()

print("{0:20}{1:20}".format("--Word--","--Stem--"))
for word in example_words:
   print ("{0:20}{1:20}".format(word,le.lemmatize(word, get_wordnet_pos(word))))
```

```python
def lemmatization(dataset):
    global le

    ## start ##
    new_dataset = [' '.join([le.lemmatize(word, get_wordnet_pos(word)) for word in word_tokenize(sentence)]) for sentence in dataset]
    ### end ###

    return new_dataset

print("Before lemmatization: ", test_sentence)
print("After lemmatization: ", lemmatization(test_sentence))
```

```python
dataset_lemmatization = lemmatization(dataset)
vocab_lemmatization = construct_vocab(dataset_lemmatization)

vocab_analysis(vocab_lemmatization)
```

### 5. Get text processing methods what you want

```python
def text_preprocessing(dataset):
    ## start ##
    dataset = lowercase(dataset)
    dataset = remove_stopwords(dataset)
    dataset = lemmatization(dataset)
    ### end ###

    return dataset
```

```python
dataset_preprocessed = text_preprocessing(dataset)
vocab = construct_vocab(dataset_preprocessed)

vocab_analysis(vocab)
```

```python
def tokenization(dataset, vocab):
    dataset_preprocessed = text_preprocessing(dataset)

    ## start ##
    dataset_tokenization = [[vocab[word] for word in word_tokenize(sentence) if word in vocab] for sentence in dataset_preprocessed]
    ### end ###

    return dataset_tokenization

example = text_preprocessing([dataset[0]])

print(example)
print([vocab[word] for word in example[0].split()])
```

# Homework: TF-IDF

### TF-IDF 구현

```python
import math
import pandas as pd
from collections import Counter
import numpy as np
from tqdm import tqdm
from typing import Dict, List


dataset_preprocessed = text_preprocessing(dataset)
vocab = construct_vocab(dataset_preprocessed)
```

```python
# 전체 단어 집합 만들기
unique_words = vocab.keys()
```

```python
list(unique_words)[:10]
```

```python
def term_frequency(data, unique_words) -> Dict[str, float]: # str: 문장 내 단어, float: 문장 내에서 단어 등장 비율
    tf = None

    ## start ##
    word_counts = Counter(word_tokenize(data))
    total_words = sum(word_counts.values())
    tf = {word: word_counts[word] / total_words for word in unique_words}
    ### end ###

    return tf
```

```python
def document_frequency(dataset) -> Dict[str, int]: # str: 전체 단어, int: dataset 내에서 단어 등장 빈도
    df_counts = None

    ## start ##
    df_counts = Counter()
    for sentence in dataset:
        unique_tokens = set(word_tokenize(sentence))
        for word in unique_tokens:                   
            df_counts[word] += 1
    ### end ###

    return df_counts
```

```python
def inverse_document_frequency(dataset, unique_words) -> Dict[str, float]: # str: 전체 단어, float: 역문서 빈도
    idf = None

    ## start ##
    N = len(dataset)
    df = document_frequency(dataset)
    idf = {word: math.log((N + 1) / (df.get(word, 1) + 1)) + 1 for word in unique_words}
    ### end ###

    return idf
```

```python
def compute_tf_idf(dataset, unique_words, idf) -> List[Dict[str, float]]: # str: 문장 내 단어, float: tf-idf 값
    tf_idf_matrix = None

    ## start ##
    tf_idf_matrix = []
    for sentence in dataset:
        tf = term_frequency(sentence, unique_words)
        tf_idf_matrix.append({word: tf.get(word, 0) * idf[word] for word in unique_words})
    ### end ###

    return tf_idf_matrix
```

```python
idf = inverse_document_frequency(dataset_preprocessed, unique_words)
tf_idf_matrix = compute_tf_idf(dataset_preprocessed, unique_words, idf)
```

```python
# DataFrame으로 변환
tf_idf_df = pd.DataFrame(tf_idf_matrix).fillna(0)

# 결과 출력
tf_idf_df
```

```python
# 특정 입력과 tf_idf dot product score가 가장 높은 리뷰 10개 출력
def get_most_similar_reviews(input, tf_idf_df, idf, top_n=10):
    input_preprocess = text_preprocessing([input])

    input_tf_idf = compute_tf_idf(input_preprocess, unique_words, idf)[0]
    input_tf_idf_df = pd.DataFrame([input_tf_idf]).fillna(0)

    dot_products = tf_idf_df.dot(input_tf_idf_df.iloc[0])
    most_similar_reviews = dot_products.sort_values(ascending=False).head(top_n)
    most_similar_reviews = [(i, dataset[i], score) for i, score in most_similar_reviews.items()]

    for i, review, score in most_similar_reviews:
        print(f'{i}\t{score}\t{review}')
```

```python
input_text = 'chicago'

get_most_similar_reviews(input_text, tf_idf_df, idf, top_n=10)
```

## Methodology
### 1. 데이터셋 개요

Cornell University에서 제공하는 Movie Review Polarity Dataset은 IMDB, Rotten Tomatoes와 같은 출처로부터 수집된 영화 리뷰 데이터로 구성된다.

- 긍정 리뷰: 5,331개 (rt-polarity.pos)
- 부정 리뷰: 5,331개 (rt-polarity.neg)
- 전체 샘플 수: 10,662개

해당 리뷰는 문장 단위의 짧은 리뷰로 구성되어 있으며, 텍스트 기반 감정 정보 또는 주제 분석에 적합하다.

### 2. 텍스트 전처리 과정

리뷰 데이터에 대해 다음과 같은 전처리를 단계적으로 수행하였다.

| 단계 | 설명 | 변화 | 
|-------|--------------|----------------|
| Lowercasing    | 모든 텍스트를 소문자로 통일       | The Rock is Destined → the rock is destined     |
| Stopword Removal     | 의미 없는 관사/전치사 제거       | the rock is destined to be → rock destined       |
| Lemmatization     | 단어를 원형으로 복원	      | players → player, playing → play         |

- nltk.word_tokenize()를 사용하여 단어 수준 토큰화를 일관성 있게 적용
- Stemming은 실험적으로 시도하였으나, 문법 파괴가 심해 최종적으로는 Lemmatization을 채택

### 3. Vocabulary 생성

- 전체 문서에 등장한 단어를 기준으로 vocab 사전 생성
- 전처리 전 단어 수: 20,298개
- 전처리 후 단어 수: 16,997개  
   → 약 16%의 노이즈 제거

### 4. TF-IDF 구현

TF-IDF는 다음 요소들로 구성되며, 직접 구현하였다.

- Term Frequency (TF): 문장 내 단어의 상대적 등장 비율
- Document Frequency (DF): 전체 문서에서 단어가 등장한 문서 수
- Inverse Document Frequency (IDF):  
  idf = log((N+1)/(df+1)) + 1 공식으로 smoothing 적용

`idf['rock'] = 1.94  # 예시`
- 각 문장마다 {단어: tf-idf score} 형태의 사전을 생성하고,
- pandas.DataFrame을 통해 10662 × 16997 크기의 희소 행렬 완성
